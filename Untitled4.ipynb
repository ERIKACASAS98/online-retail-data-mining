# Punto 2 — Segmentación de clientes con RFM y k-means

Según el EC3, una tarea es **descriptiva** cuando no existe una variable objetivo que predecir y el propósito es descubrir la estructura interna de los datos. El **agrupamiento** o *clustering* es su forma más común: separa los registros en grupos de modo que los miembros de un mismo grupo se parezcan entre sí y se diferencien de los demás.

Aquí surge una dificultad que no existía en los talleres anteriores: **no hay variables listas para agrupar**. El conjunto tiene líneas de factura, no clientes. Hay que construir una descripción de cada cliente a partir de su historial, y el estándar de la industria para hacerlo es el modelo **RFM**:

- **Recencia**: cuántos días han pasado desde su última compra. Menos es mejor.
- **Frecuencia**: cuántas facturas distintas ha generado. Más es mejor.
- **Monto**: cuánto ha gastado en total. Más es mejor.

> **Enunciado.** Construya la tabla RFM, prepare los datos para el agrupamiento, determine un número razonable de grupos, aplique k-means e interprete cada segmento en términos de negocio.

```python
# Comparación de facturación

df["Importe_sin_depurar"] = (
    df["Quantity"] * df["Price"]
)

facturacion_bruta = df["Importe_sin_depurar"].sum()
facturacion_limpia = df_clean["Importe"].sum()

print(
    f"Facturación sin depurar: "
    f"£{facturacion_bruta:,.2f}"
)

print(
    f"Facturación depurada: "
    f"£{facturacion_limpia:,.2f}"
)

print(
    f"Diferencia: "
    f"£{facturacion_limpia - facturacion_bruta:,.2f}"
)
```

```python
# 2.1 Tabla RFM

fecha_referencia = (
    df_clean["InvoiceDate"].max()
    + pd.Timedelta(days=1)
)

print("Fecha de referencia:", fecha_referencia)

rfm = (
    df_clean
    .groupby("CustomerID")
    .agg(
        Recencia=(
            "InvoiceDate",
            lambda x: (fecha_referencia - x.max()).days
        ),
        Frecuencia=("Invoice", "nunique"),
        Monto=("Importe", "sum")
    )
    .reset_index()
)

print("Número de clientes:", len(rfm))

display(rfm.head())

display(
    rfm[
        ["Recencia", "Frecuencia", "Monto"]
    ].describe()
)
```

```python
# 2.2 Histogramas RFM

variables = [
    "Recencia",
    "Frecuencia",
    "Monto"
]

for variable in variables:
    plt.figure(figsize=(7, 4))
    plt.hist(rfm[variable], bins=40)
    plt.title(f"Distribución de {variable}")
    plt.xlabel(variable)
    plt.ylabel("Frecuencia")
    plt.show()
```

```python
print("=== ASIMETRÍA ===")

print(
    rfm[
        ["Recencia", "Frecuencia", "Monto"]
    ].skew()
)
```

```python
from sklearn.preprocessing import StandardScaler

# Trabajamos con copia
rfm_modelo = rfm[
    ["Recencia", "Frecuencia", "Monto"]
].copy()

# Transformación logarítmica
rfm_log = np.log1p(rfm_modelo)

print("Asimetría después del logaritmo:")
print(rfm_log.skew())
```

```python
for variable in rfm_log.columns:
    plt.figure(figsize=(7, 4))
    plt.hist(rfm_log[variable], bins=40)
    plt.title(f"{variable} después de log1p")
    plt.xlabel(variable)
    plt.ylabel("Frecuencia")
    plt.show()
```

```python
scaler = StandardScaler()

rfm_scaled = scaler.fit_transform(rfm_log)

rfm_scaled = pd.DataFrame(
    rfm_scaled,
    columns=[
        "Recencia",
        "Frecuencia",
        "Monto"
    ],
    index=rfm.index
)

display(rfm_scaled.head())

print("\nMedias:")
print(rfm_scaled.mean().round(4))

print("\nDesviaciones estándar:")
print(rfm_scaled.std().round(4))
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

inercias = []
siluetas = []

ks = range(2, 9)

for k in ks:

    modelo = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    etiquetas = modelo.fit_predict(rfm_scaled)

    inercias.append(modelo.inertia_)

    siluetas.append(
        silhouette_score(
            rfm_scaled,
            etiquetas
        )
    )
```

```python
resultados_k = pd.DataFrame({
    "k": list(ks),
    "Inercia": inercias,
    "Silueta": siluetas
})

display(resultados_k)
```

```python
plt.figure(figsize=(7, 4))

plt.plot(
    list(ks),
    inercias,
    marker="o"
)

plt.xlabel("Número de clusters (k)")
plt.ylabel("Inercia")
plt.title("Método del codo")

plt.show()
```

```python
plt.figure(figsize=(7, 4))

plt.plot(
    list(ks),
    siluetas,
    marker="o"
)

plt.xlabel("Número de clusters (k)")
plt.ylabel("Silhouette score")
plt.title("Coeficiente de silueta")

plt.show()
```

```python
mejor_k_silueta = resultados_k.loc[
    resultados_k["Silueta"].idxmax(),
    "k"
]

print(
    "k con mayor silhouette score:",
    mejor_k_silueta
)
```

```python

k_elegido = 4

kmeans = KMeans(
    n_clusters=k_elegido,
    random_state=42,
    n_init=10
)

rfm["Cluster"] = kmeans.fit_predict(
    rfm_scaled
)

display(rfm.head())
```

```python
perfil_clusters = (
    rfm
    .groupby("Cluster")
    .agg(
        Clientes=("CustomerID", "count"),
        Recencia_media=("Recencia", "mean"),
        Frecuencia_media=("Frecuencia", "mean"),
        Monto_medio=("Monto", "mean"),
        Facturacion_total=("Monto", "sum")
    )
)

perfil_clusters["Porcentaje_clientes"] = (
    perfil_clusters["Clientes"]
    / perfil_clusters["Clientes"].sum()
    * 100
)

perfil_clusters["Porcentaje_facturacion"] = (
    perfil_clusters["Facturacion_total"]
    / perfil_clusters["Facturacion_total"].sum()
    * 100
)

display(
    perfil_clusters.round(2)
)
```

# Punto 3 — Reglas de asociación: qué se compra junto

El **análisis de canasta de compra** busca patrones del tipo "quien compra A también tiende a comprar B". Sus tres medidas, definidas en el EC3, son:

- **Soporte**: qué proporción de todas las facturas contiene el conjunto de productos. Mide qué tan frecuente es la combinación.
- **Confianza**: de las facturas que contienen A, qué proporción contiene también B. Mide qué tan fiable es la regla.
- **Lift**: cuántas veces más probable es que aparezca B cuando está A, frente a lo esperable si fueran independientes. Un lift de 1 significa ninguna asociación; mayor que 1, asociación positiva.

El algoritmo **Apriori** busca primero los conjuntos frecuentes y luego deriva las reglas. Su dificultad es combinatoria: con 4.619 productos, el número de combinaciones posibles es astronómico, y por eso se exige un soporte mínimo que pode el espacio de búsqueda. Esta es la razón por la cual el análisis de canasta **no se puede hacer a mano ni en una hoja de cálculo**.

```python
# 3.1 Subconjunto Reino Unido, año 2011

df_market = df_clean[
    (df_clean["Country"] == "United Kingdom") &
    (df_clean["InvoiceDate"].dt.year == 2011)
].copy()

print(
    "Registros UK 2011:",
    f"{len(df_market):,}"
)

print(
    "Facturas:",
    f"{df_market['Invoice'].nunique():,}"
)

print(
    "Productos distintos:",
    f"{df_market['Description'].nunique():,}"
)
```

```python
top150 = (
    df_market
    .groupby("Description")["Invoice"]
    .nunique()
    .sort_values(ascending=False)
    .head(150)
    .index
)

df_market150 = df_market[
    df_market["Description"].isin(top150)
].copy()

print(
    "Registros después de seleccionar top 150:",
    f"{len(df_market150):,}"
)

print(
    "Productos:",
    df_market150["Description"].nunique()
)
```

```python
basket = (
    df_market150
    .groupby(
        ["Invoice", "Description"]
    )["Quantity"]
    .sum()
    .unstack()
    .fillna(0)
)

# Convertimos a presencia/ausencia
basket = basket > 0

print(
    "Dimensiones de la matriz:",
    basket.shape
)

display(basket.head())
```

```python
import warnings

warnings.filterwarnings("ignore", category=DeprecationWarning)
```

```python
from mlxtend.frequent_patterns import (
    apriori,
    association_rules
)

frequent_itemsets = apriori(
    basket,
    min_support=0.02,
    use_colnames=True
)

print(
    "Conjuntos frecuentes encontrados:",
    len(frequent_itemsets)
)

display(
    frequent_itemsets
    .sort_values(
        "support",
        ascending=False
    )
    .head(20)
)
```

```python
reglas = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.2
)

print(
    "Reglas generadas:",
    len(reglas)
)
```

```python
reglas_filtradas = reglas[
    (reglas["confidence"] >= 0.30) &
    (reglas["lift"] > 1)
].copy()

reglas_filtradas = (
    reglas_filtradas
    .sort_values(
        ["lift", "confidence"],
        ascending=False
    )
)

print(
    "Reglas después del filtro:",
    len(reglas_filtradas)
)
```

```python
def convertir_itemset(x):
    return ", ".join(list(x))

reglas_mostrar = reglas_filtradas[
    [
        "antecedents",
        "consequents",
        "support",
        "confidence",
        "lift"
    ]
].copy()

reglas_mostrar["antecedents"] = (
    reglas_mostrar["antecedents"]
    .apply(convertir_itemset)
)

reglas_mostrar["consequents"] = (
    reglas_mostrar["consequents"]
    .apply(convertir_itemset)
)

display(
    reglas_mostrar.head(20)
)
```

```python
display(
    reglas_mostrar
    .sort_values(
        "lift",
        ascending=False
    )
    .head(10)
    .round({
        "support": 4,
        "confidence": 4,
        "lift": 2
    })
)
```

Los resultados muestran que productos de **colecciones o diseños relacionados tienden a comprarse juntos**. Por ello, la tienda podría crear paquetes de productos, promociones cruzadas y recomendaciones de “comprados juntos” para aumentar el valor de cada compra.
